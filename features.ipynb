```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# --- the features CSV to plot (one file's complete-track cells) ---
FEATURES_CSV = r"C:\Users\ACID\Desktop\DENV\programs\batch_output\A09.1_pos29_sc_gaussian8_eq(A07.3_avg41pos_x81t_sc_gaussian8)_med2389\results\A09.1_pos29_sc_gaussian8_eq(A07.3_avg41pos_x81t_sc_gaussian8)_med2389_features_complete81.csv"

df = pd.read_csv(FEATURES_CSV)

# where to save the per-feature plots
stem = os.path.splitext(os.path.basename(FEATURES_CSV))[0]
plot_dir = os.path.join(os.path.dirname(FEATURES_CSV), f"{stem}_plots")
os.makedirs(plot_dir, exist_ok=True)

# feature columns — skip ids / coordinates
SKIP = {"FrameID", "CellID", "ROI_filename", "x", "y"}
feature_cols = [c for c in df.columns if c not in SKIP]
print(f"{len(feature_cols)} features, {df['CellID'].nunique()} cells")

for feat in feature_cols:
    fig, ax = plt.subplots(figsize=(9, 5))
    cell_ids = sorted(df["CellID"].unique())
    colors = cm.tab20(np.linspace(0, 1, len(cell_ids)))

    for color, cell_id in zip(colors, cell_ids):
        g = df[df["CellID"] == cell_id].sort_values("FrameID")
        ax.plot(g["FrameID"], g[feat], alpha=0.7, linewidth=1, color=color)
        # write the CellID at the line's last point
        last = g.iloc[-1]
        ax.annotate(str(cell_id), (last["FrameID"], last[feat]),
                    fontsize=6, color=color, va="center",
                    xytext=(3, 0), textcoords="offset points")

    ax.set_xlabel("Frame (t)")
    ax.set_ylabel(feat)
    ax.set_title(feat, fontweight="bold")
    ax.spines[["top", "right"]].set_visible(False)
    fig.tight_layout()
    fig.savefig(os.path.join(plot_dir, f"{feat}.png"), dpi=120, bbox_inches="tight")
    plt.close(fig)

print(f"saved {len(feature_cols)} plots -> {plot_dir}")
```
